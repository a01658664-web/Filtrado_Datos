## FILTROS

**A. Los registros de 5 “host name” diferentes**

```python
# Quitamos los nombres vacíos y repetidos
# y mostramos los primeros 5

filtroA = df.dropna(subset=["host_name"]).drop_duplicates("host_name").head(6)

# Guardamos el resultado
filtroA.to_csv("IncisoA.csv")

# Mostramos el resultado
print(filtroA)
```

**B. Los registros de los host que se hallan unido a Airbnb “despues del año 2020”**

```python
# Convertimos los datos de last_review a fecha

df['last_review'] = pd.to_datetime(df['last_review'])

# Buscamos las fechas que sean después del año 2020

filtroB = df[df["last_review"] > '31-12-2020']

# Creamos un archivo con los resultados

filtroB.to_csv("IncisoB.csv")

# Vemos el resultado

filtroB
```

**D. Los registros de las filas “Múltiplos de 200… (400, 600, 800, etc.)”**

```python
# Tomamos una fila cada 200 registros, empezando desde la fila 0
filtroD = df[::200]

# Guardamos los datos en un archivo CSV
filtroD.to_csv("IncisoD.csv")

# Mostramos los resultados
print(filtroD)
```

**E. Los registros de los tipos de cuarto: “Entire home” y “Private room”**

```python
# Seleccionamos los registros que tengan alguno de los dos tipos de habitación
filtroE = df[df["room_type"].isin(["Entire home/apt", "Private room"])]

# Guardamos los resultados en un archivo CSV
filtroE.to_csv("IncisoE.csv")

# Mostramos los resultados
print(filtroE)
```
